```python
import re, unicodedata

def clean_text(s):
    s = str(s)
    s = ''.join(ch for ch in s if unicodedata.category(ch) != 'Cf')
    return s.strip()

score_pattern = re.compile(r'\d+\s*[–\-−]\s*\d+')

group_matches = []
match_num = 0

for i in range(11, 94):
    t = tables[i]
    if t.shape != (1, 3):
        continue
    cols = [clean_text(c) for c in t.columns.tolist()]
    if score_pattern.search(cols[1]):
        match_num += 1
        row = t.iloc[0]
        group_matches.append({
            "رقم_المباراة": match_num,
            "الفريق1": cols[0],
            "النتيجة": cols[1],
            "الفريق2": cols[2],
            "هدافي_الفريق1": row.iloc[0],
            "هدافي_الفريق2": row.iloc[2],
            "الدور": "دور المجموعات",
        })

matches_df = pd.DataFrame(group_matches)
print("عدد المباريات:", matches_df.shape)
matches_df.head(15)
```

```python
knockout_matches = []
ko_match_num = 0

# اسماء الأدوار بالترتيب حسب مكان كل مجموعة مباريات في الجدول
# (هنحطها بسيطة الأول، ونظبطها بعدين لو محتاجة تعديل)
for i in range(96, 128):
    t = tables[i]
    cols = [clean_text(c) for c in t.columns.tolist()]

    if t.shape[0] == 1 and t.shape[1] == 3:
        # مباراة عادية (90 دقيقة أو وقت إضافي بدون ركلات ترجيح)
        if score_pattern.search(cols[1]):
            ko_match_num += 1
            row = t.iloc[0]
            knockout_matches.append({
                "رقم_المباراة": ko_match_num,
                "الفريق1": cols[0],
                "النتيجة": cols[1],
                "الفريق2": cols[2],
                "هدافي_الفريق1": row.iloc[0],
                "هدافي_الفريق2": row.iloc[2],
                "index_جدول": i
            })
    else:
        # مباراة اتحسمت بركلات ترجيح (شكل الجدول مختلف - 3 صفوف)
        knockout_matches.append({
            "رقم_المباراة": None,
            "الفريق1": cols[0] if len(cols) > 0 else None,
            "النتيجة": cols[1] if len(cols) > 1 else None,
            "الفريق2": cols[2] if len(cols) > 2 else None,
            "هدافي_الفريق1": "راجع الجدول يدويًا",
            "هدافي_الفريق2": "راجع الجدول يدويًا",
            "index_جدول": i
        })

knockout_df = pd.DataFrame(knockout_matches)
print(knockout_df.shape)
knockout_df
```

```python
import re

def parse_scorers(text, team, match_id):
    if pd.isna(text):
        return []
    text = str(text).replace('\xa0', ' ')
    pattern = re.compile(r'([\u0600-\u06FF][\u0600-\u06FF\.\s]*?)\s*((?:\d+(?:\+\d+)?\'(?:\s*\([^)]*\))?(?:\s*،\s*)?)+)')
    results = []
    for name, times_blob in pattern.findall(text):
        name = name.strip()
        own_goal = 'هـ.ذ.' in times_blob
        is_penalty = 'ر.ج.' in times_blob
        times = re.findall(r"\d+(?:\+\d+)?'", times_blob)
        for t in times:
            results.append({
                "رقم_المباراة": match_id,
                "اللاعب": name,
                "الفريق_المستفيد": team,
                "الدقيقة": t,
                "هدف_عكسي": own_goal,
            })
    return results

all_scorers = []
for _, row in all_matches_df.iterrows():
    all_scorers += parse_scorers(row["هدافي_الفريق1"], row["الفريق1"], row["رقم_المباراة"])
    all_scorers += parse_scorers(row["هدافي_الفريق2"], row["الفريق2"], row["رقم_المباراة"])

scorers_df = pd.DataFrame(all_scorers)
print("إجمالي الأهداف المستخرجة:", scorers_df.shape)

# ترتيب الهدافين (نستبعد الأهداف العكسية لأنها مش لصالح لاعب الفريق)
top_scorers = (
    scorers_df[~scorers_df["هدف_عكسي"]]
    .groupby("اللاعب")
    .size()
    .reset_index(name="عدد_الأهداف")
    .sort_values("عدد_الأهداف", ascending=False)
    .reset_index(drop=True)
)

print(top_scorers.head(15))

# تصدير
scorers_df.to_csv("world_cup_2026_all_goals.csv", index=False, encoding="utf-8-sig")
top_scorers.to_csv("world_cup_2026_top_scorers.csv", index=False, encoding="utf-8-sig")

from google.colab import files
files.download("world_cup_2026_all_goals.csv")
files.download("world_cup_2026_top_scorers.csv")
```
