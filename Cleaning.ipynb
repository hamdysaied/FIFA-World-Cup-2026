```python
import pandas as pd

df = pd.read_csv("world_cup_2026_full_dataset.csv")

# استبدال النولز بقيمة واضحة في الأعمدة الخمسة اللي فيها نولز منطقية
cols_to_fill = [
    "هدافي_الفريق1",
    "هدافي_الفريق2",
    "ركلات_ترجيح_الفريق1",
    "نتيجة_الركلات",
    "ركلات_ترجيح_الفريق2",
]

df[cols_to_fill] = df[cols_to_fill].fillna("لا يوجد")

# تأكيد إن مفيش أي نولز باقية في الملف كله
print("عدد النولز الكلي بعد التنظيف:", df.isna().sum().sum())

df.head(10)
```

```python
df.to_csv("world_cup_2026_full_dataset_clean.csv", index=False, encoding="utf-8-sig")

from google.colab import files
files.download("world_cup_2026_full_dataset_clean.csv")
```
